```python
import re
import json

sample_json_data = [
    {"title": "Samsung Galaxy Fold and Smartphone Review", "textBody": "Samsung introduced the new Galaxy smartphone with advanced display technology and chip.", "source": "news"},
    {"title": "Galaxy Watch and Mobile Devices Market", "textBody": "Samsung Galaxy watch features new battery life and screen quality for mobile market.", "source": "tech"}
]

with open("samsung.json", "w", encoding="utf-8") as f:
    json.dump(sample_json_data, f, ensure_ascii=False)

stop_words_list = ["AND", "THE", "WITH", "FOR", "NEW", "IN", "ON", "AT", "OF", "TO", "IS"]
with open("stop.txt", "w", encoding="utf-8") as f:
    f.write("\n".join(stop_words_list))


with open("samsung.json", "r", encoding="utf-8") as f:
    raw_text = f.read()

data = json.loads(raw_text)
combined_text = ""
for doc in data:
    title = doc.get("title", "")
    body = doc.get("textBody", "")
    combined_text += f" {title} {body}"

clean_text = re.sub(r'[^A-Za-z0-9\s]', ' ', combined_text)
clean_text = clean_text.upper()
clean_text = re.sub(r'\b\w{1,2}\b', ' ', clean_text)
clean_text = re.sub(r'\s+', ' ', clean_text)

words = clean_text.strip().split(' ')
words.sort()

word_counts = {}
for word in words:
    if word:
        word_counts[word] = word_counts.get(word, 0) + 1

sorted_dict = dict(sorted(word_counts.items(), key=lambda item: item[1], reverse=True))

print(" Топ-10 слів до видалення стоп-слів:")
for i, (w, count) in enumerate(list(sorted_dict.items())[:10], 1):
    print(f"{i}. {w}: {count}")

with open("stop.txt", "r", encoding="utf-8") as f:
    stop_text = f.read().upper()

stop_words = stop_text.split('\n')

M = 20
filtered_words = []

for word, count in sorted_dict.items():
    if word not in stop_words:
        filtered_words.append((word, count))
    if len(filtered_words) >= M:
        break

print(f"\n Top-{M} найвагоміших слів (після видалення стоп-слів):")
for j, (word, count) in enumerate(filtered_words, 1):
    print(f"{j}. {word} — {count}")

with open("words.txt", "w", encoding="utf-8") as f:
    for word, count in filtered_words:
        f.write(f"{count} {word}\n")

print("\n Результати успішно збережено у файл words.txt!")
```
